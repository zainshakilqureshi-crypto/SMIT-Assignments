## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
# Write your code below
# Student Marks Processor

def process_marks(marks):
    highest = marks[0]
    lowest = marks[0]
    total = 0
    
    for m in marks:
        if m > highest:
            highest = m
        if m < lowest:
            lowest = m
        total += m
    
    average = total / len(marks)
    
    return highest, lowest, total, average

student_marks = [85, 92, 78, 90, 88, 76, 95]

high, low, total_marks, avg_marks = process_marks(student_marks)

print(f"Highest Mark: {high}")
print(f"Lowest Mark: {low}")
print(f"Total Marks: {total_marks}")
print(f"Average Marks: {avg_marks:.2f}")
````

Build a very simple banking system that keeps track of an account balance.

  - `deposit(amount)`
  - `withdraw(amount)`
  - `check_balance()`
- The withdraw function must **not** allow withdrawal if the amount is greater than the available balance.
- Demonstrate deposit, withdraw, and checking balance.

**Focus:** `global` + multiple functions + conditions

````python
# Write your code below
# Bank Balance System

# Global variable for account balance
account_balance = 10000  # initial amount

def deposit(amount):
    global account_balance
    account_balance += amount
    print(f"Rs. {amount} deposited. New balance: Rs. {account_balance}")

def withdraw(amount):
    global account_balance
    if amount > account_balance:
        print(f"Withdrawal failed! Not enough balance. Available: Rs. {account_balance}")
    else:
        account_balance -= amount
        print(f"Rs. {amount} withdrawn. Remaining balance: Rs. {account_balance}")

def check_balance():
    print(f"Current Balance: Rs. {account_balance}")

# Demonstrate
print("--- Banking System Demo ---")
check_balance()

print("\n1. Depositing 5000:")
deposit(5000)

print("\n2. Withdrawing 3000:")
withdraw(3000)

print("\n3. Trying to withdraw 20000 (more than balance):")
withdraw(20000)

print("\n4. Final check:")
check_balance()
````

## Question 23 — Nested Salary Calculator

**Problem:**  
Salary calculation has several smaller steps. Use nested functions to organize the logic cleanly.

**Requirements:**
- Create an outer function called `calculate_salary()`.
- Inside it, create separate **nested functions** for:
  - Calculating allowance
  - Calculating deduction
  - Calculating final salary
- The outer function should combine the results and return the final salary.
- Call the outer function and display the result.

**Focus:** Nested functions

````python
# Write your code below
# Nested Salary Calculator

def calculate_salary(basic):
    
    def calculate_allowance():
        return basic * 0.20
            
    def calculate_deduction():
        return basic * 0.10

    def calculate_final():
        allowance = calculate_allowance()
        deduction = calculate_deduction()
        final = basic + allowance - deduction
        return final, allowance, deduction

    final_salary, allowance, deduction = calculate_final()
    return final_salary, allowance, deduction

basic = 50000
final, allowance, deduction = calculate_salary(basic)

print(f"Basic Salary: Rs. {basic}")
print(f"Allowance (20%): Rs. {allowance}")
print(f"Deduction (10%): Rs. {deduction}")
print(f"Final Salary: Rs. {final}")
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
# Write your code below
# Nested Order Processing

def process_order(price, quantity):
    
    def calculate_subtotal():
        return price * quantity

    def calculate_discount(subtotal):
        if subtotal > 5000:
            return subtotal * 0.10
        else:
            return 0
            
    def calculate_tax(amount_after_discount):
        return amount_after_discount * 0.05

    def calculate_final():
        subtotal = calculate_subtotal()
        discount = calculate_discount(subtotal)
        after_discount = subtotal - discount
        tax = calculate_tax(after_discount)
        final = after_discount + tax
        return subtotal, discount, tax, final

    subtotal, discount, tax, final = calculate_final()
    return subtotal, discount, tax, final

print("--- Order 1 ---")
subtotal, discount, tax, final = process_order(2000, 3)
print(f"Subtotal: Rs. {subtotal}")
print(f"Discount: Rs. {discount}")
print(f"Tax (5%): Rs. {tax}")
print(f"Final Amount: Rs. {final}")

print("\n--- Order 2 ---")
subtotal, discount, tax, final = process_order(1000, 2)
print(f"Subtotal: Rs. {subtotal}")
print(f"Discount: Rs. {discount}")
print(f"Tax (5%): Rs. {tax}")
print(f"Final Amount: Rs. {final}")
````

## Question 26 — Lambda With Student Marks

**Problem:**  
You have a list of student marks. You need to process them quickly using lambda.

**Requirements:**
- Given a list of student marks, use a **lambda** function (with `map` or list comprehension) to create a new list containing the **square** of each mark.
- Create another lambda-based operation that checks whether a mark is **even** or not.
- Display both results clearly.

**Focus:** Lambda + list processing

````python
# Write your code below
# Lambda With Student Marks

marks = [45, 60, 75, 80, 91, 55]

squared_marks = list(map(lambda x: x ** 2, marks))

even_check = list(map(lambda x: x % 2 == 0, marks))

print("Original Marks:", marks)
print("Squared Marks:", squared_marks)
print("Is Even?:", even_check)

print("\n--- Detail ---")
for mark, square, is_even in zip(marks, squared_marks, even_check):
    status = "Even" if is_even else "Odd"
    print(f"Mark: {mark} -> Square: {square} -> {status}")
````

````python
# Write your code below
# Shopping Billing System

# Function 1: Calculate individual product total
def product_total(price, qty):
    return price * qty

# Function 2: Calculate overall subtotal using *args (Flexible argument design)
def calculate_subtotal(*totals):
    subtotal = sum(totals)
    return subtotal

# Function 3: Calculate discount - 10% if subtotal > 5000
def calculate_discount(subtotal):
    if subtotal > 5000:
        return subtotal * 0.10
    else:
        return 0

# Function 4: Calculate tax - 5% on amount after discount
def calculate_tax(amount_after_discount):
    return amount_after_discount * 0.05

# Function 5: Calculate final bill
def calculate_final_bill(subtotal, discount, tax):
    return subtotal - discount + tax

# Main Program
# Hard-coded products: (name, price, quantity)
products = [
    ("Shampoo", 500, 2),
    ("Soap", 100, 5),
    ("Toothpaste", 250, 3),
    ("Oil", 800, 1)
]

print("========== SHOPPING BILL ==========")
print(f"{'Product':<12} {'Price':<7} {'Qty':<5} {'Total'}")
print("-----------------------------------")

individual_totals = []
for name, price, qty in products:
    total = product_total(price, qty)
    individual_totals.append(total)
    print(f"{name:<12} {price:<7} {qty:<5} {total}")

# Using *args - unpack list
subtotal = calculate_subtotal(*individual_totals)
discount = calculate_discount(subtotal)
after_discount = subtotal - discount
tax = calculate_tax(after_discount)
final = calculate_final_bill(subtotal, discount, tax)

print("-----------------------------------")
print(f"Overall Subtotal : Rs. {subtotal}")
print(f"Discount         : Rs. {discount}")
print(f"Tax (5%)         : Rs. {tax}")
print("-----------------------------------")
print(f"FINAL BILL       : Rs. {final}")
print("===================================")
````

Create a complete student result system using functions.

- Do **not** write the entire program inside one function

````python
# Write your code below
# Student Result Management System

def accept_student_info(**kwargs):
    return kwargs

def accept_marks(*args):
    marks_list = list(args) 
    return marks_list

def calculate_total(marks):
    total = 0
    for m in marks:  
        total += m
    return total

def calculate_percentage(total, num_subjects):
    percentage = (total / (num_subjects * 100)) * 100
    return percentage

def determine_grade(percentage):
    get_grade = lambda p: "A+" if p >= 80 else "A" if p >= 70 else "B" if p >= 60 else "C" if p >= 50 else "D" if p >= 40 else "F"
    return get_grade(percentage)

def determine_result(marks):
    for mark in marks:  # Loop required
        if mark < 40:  # Condition required
            return "Fail"
    return "Pass"

def display_result_card(info, marks, total, percentage, grade, result):
    print("\n========== STUDENT RESULT CARD ==========")
    for key, value in info.items(): # Loop
        print(f"{key}: {value}")
    print("------------------------------------------")
    print(f"Marks List: {marks}")
    print(f"Total Marks: {total} / {len(marks)*100}")
    print(f"Percentage: {percentage:.2f}%")
    print(f"Grade: {grade}")
    print(f"Result: {result}")
    print("==========================================\n")


student = accept_student_info(Name="Ahmed Ali", Roll_No="SMIT-101", Class="Python", City="Karachi")
marks = accept_marks(85, 78, 92, 88, 75) # List will be created inside

total = calculate_total(marks)
per = calculate_percentage(total, len(marks))

grade = determine_grade(per)
final_result = determine_result(marks)

display_result_card(student, marks, total, per, grade, final_result)
````

**Challenge:**  
Design the program yourself. Decide:

This question tests your **complete understanding** of Python functions.

**Focus:** Complete modular real-world system

````python
# Write your code below
# FINAL CHALLENGE 
def customer_info(**kwargs): return kwargs
def add_products(*products): return list(products)
def calculate_item_total(price, quantity): return price * quantity

def calculate_subtotal(product_list):
    subtotal = 0
    for name, price, qty in product_list:
        subtotal += calculate_item_total(price, qty)
    return subtotal

def calculate_discount(subtotal, discount_rate=0.10):
    if subtotal > 50000: return subtotal * discount_rate
    return 0

def calculate_tax(amount, tax_rate=0.05): return amount * tax_rate

def delivery_charges(subtotal, charge=500):
    if subtotal > 30000: return 0
    return charge

def calculate_final_bill(subtotal):
    def apply_all_charges():
        discount = calculate_discount(subtotal)
        after_discount = subtotal - discount
        tax = calculate_tax(after_discount)
        delivery = delivery_charges(subtotal)
        final = after_discount + tax + delivery
        return discount, tax, delivery, final
    return apply_all_charges()

def display_order_summary(customer, products, subtotal, discount, tax, delivery, final_bill):
    print("\n========== ORDER SUMMARY ==========")
    for k, v in customer.items(): print(f"{k}: {v}")
    for name, price, qty in products:
        print(f"{name} | {price} x {qty} = {calculate_item_total(price, qty)}")
    print(f"Subtotal: {subtotal}, Discount: {discount:.2f}, Tax: {tax:.2f}, Delivery: {delivery}")
    print(f"FINAL BILL: {final_bill:.2f}")
    print("===================================\n")

invoice_title = lambda name: f"Invoice for {name.upper()}"

cust = customer_info(Name="Ali Ahmed", Phone="0300-1234567", City="Karachi")
cart = add_products(("HP Laptop", 80000, 1), ("Mouse", 2500, 2), ("Keyboard", 3500, 1))
sub = calculate_subtotal(cart)
disc, tx, deli, final = calculate_final_bill(sub)
print(invoice_title(cust["Name"]))
display_order_summary(cust, cart, sub, disc, tx, deli, final)
````

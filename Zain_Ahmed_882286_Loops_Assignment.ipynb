## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0
account_created = False

while True:
    print("\n--- Banking System ---")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        if account_created:
            print("Account already created!")
        else:
            account_created = True
            balance = 0
            print("Account Created Successfully!")

    elif choice == "2":
        if not account_created:
            print("Create account first!")
        else:
            amount = int(input("Enter deposit amount: "))
            if amount > 0:
                balance += amount
                print(f"Deposited {amount}")
            else:
                print("Invalid amount")

    elif choice == "3":
        if not account_created:
            print("Create account first!")
        else:
            amount = int(input("Enter withdraw amount: "))
            if amount <= balance:
                balance -= amount
                print(f"Withdrawn {amount}")
            else:
                print("Insufficient balance!")

    elif choice == "4":
        if not account_created:
            print("Create account first!")
        else:
            print(f"Your Balance = {balance}")

    elif choice == "5":
        print("Exiting...")
        break
    else:
        print("Invalid choice")
```

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
students = [
    [80, 90, 70, 85],
    [50, 32, 45, 40]
]

def get_grade(total):
    avg = total / 4
    if avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 50:
        return "D"
    else:
        return "F"

for i in range(len(students)):
    marks = students[i]
    failed = False
    
    # Inner loop -> check each subject
    for j in range(len(marks)):
        if marks[j] < 33:
            print(f"Student {i+1}: {marks} -> Failed in Subject {j+1}")
            failed = True
            break
    
    if not failed:
        total = sum(marks)
        grade = get_grade(total)
        print(f"Student {i+1}: {marks} -> Grade: {grade}")
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
packages = [
    [4, 5, 5],  # Package 1
    [3, 4, 0]   # Package 2
]

for i in range(len(packages)):
    ratings = packages[i]
    total = 0
    count = 0

    # Inner loop -> 3 features
    for rating in ratings:
        if rating == 0:
            continue  # Skip if rating = 0
        total += rating
        count += 1

    avg = total / count if count > 0 else 0
    print(f"Package {i+1}: {ratings} -> Avg = {round(avg, 2)}")
```

* If 81–100 → Fine = 2000

*If rainy weather AND speed > 60 → fine increases by 50%

```python
Speed = int(input("Enter Speed: "))
is_Emergency = input("Emergency Vehicle? (yes / no): ")
weather = input("weather(rainy / normal): ")

fine = 0

if is_Emergency == "yes":
    print("No fine - Emergency Vehicle")
else:
    if Speed <= 60:
        print("No fine")
    elif Speed <= 80:
        fine = 500
        if weather == "rainy":
            fine = fine * 1.5
        print(f"fine = {int(fine)}")
    elif Speed <= 100:
        fine = 2000
        if weather == "rainy":
            fine = fine * 1.5
        print(f"fine = {int(fine)}")
    else:
        ("License Suspended")
              
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
amount = int(input("Transaction amount: "))
location = input("Location match? (Yes/No): ")
time = input("Time of transaction (Day/Night): ")
prev_count = int(input("Previous transactions count in 1 hour: "))
device = input("Device verified? (Yes/No): ")

# Rule 1: Many transactions
if prev_count > 10:
    print("Freeze Account")

# Rule 2: Location mismatch + high amount
elif location == "No" and amount > 50000:  # high amount ka matlab 50k+ le lete hain
    print("Fraud Alert")

# Rule 3: Night + unverified device
elif time == "Night" and device == "No":
    print("High Risk")

# Rule 4: High amount
elif amount > 100000:
    print("Suspicious Transaction")

else:
    print("Normal Transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal = input("Criminal record? (Yes/No): ")
cases = int(input("Number of cases: "))
travel = input("Travel history flagged? (Yes/No): ")
financial = int(input("Financial activity score (0-100): "))
last_days = int(input("Last activity days ago: "))

# Rule 1: criminal Yes and cases > 3 -> High Risk
if criminal == "Yes" and cases > 3:
    print("High Risk")

# Rule 2: flagged travel + suspicious financial (score < 40 ko suspicious lete hain)
elif travel == "Yes" and financial < 40:
    print("High Risk")

# Rule 3: no activity long time
elif last_days > 365:
    print("Monitor")

# Rule 4: clean record + normal activity
elif criminal == "No" and cases == 0 and travel == "No" and financial >= 60:
    print("Low Risk")

# Rule 5: mixed signals
else:
    print("Medium Risk")
```
